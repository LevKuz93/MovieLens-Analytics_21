## ⭐ Класс `Ratings`: Анализ пользовательских оценок

> **Назначение:** Агрегация и статистический анализ рейтингов из `ratings.csv`  
> **Источник:** `./ml-latest-small/ratings.csv` + `movies.csv` (для названий фильмов)  
> **Архитектура:** Основной класс + два вложенных класса для разделения логики

---

### 🏗️ Архитектура класса `Ratings`

| Уровень | Класс / Метод | Назначение |
|:--------|:--------------|:-----------|
| 🔹 Основной | `Ratings` | Загрузка данных, лимит выборки |
|  | `read_csv_to_dicts()` | Парсинг CSV в список словарей |
| 🎬 Вложенный | `Ratings.Movies` | Анализ агрегированных данных **по фильмам** |
|  | `dist_by_year()` | Распределение оценок по годам |
|  | `dist_by_rating()` | Распределение по значениям рейтинга |
|  | `top_by_num_of_ratings()` | Топ фильмов по количеству оценок |
|  | `top_by_ratings()` | Топ по среднему/медианному рейтингу |
|  | `top_controversial()` | Топ по дисперсии оценок |
|  | `data_movie_id_title()` | Словарь `movieId → title` |
|  | `average()` / `median()` / `variance()` | Статистические утилиты |
| 👥 Вложенный | `Ratings.Users` *(наследуется от `Movies`)* | Анализ агрегированных данных **по пользователям** |
|  | `dist_by_num_of_rating()` | Распределение пользователей по активности |
|  | `dist_by_rating_values()` | Распределение по среднему/медианному рейтингу |
|  | `top_controversial_users()` | Топ пользователей с наибольшим разбросом |

#
> 💡 **Почему вложенные классы?**  
> - `Ratings.Movies` и `Ratings.Users` имеют общий доступ к данным через `outer`  
> - Наследование позволяет переиспользовать статистические методы (`average`, `median`, `variance`)  
> - Чёткое разделение: «анализ фильмов» ≠ «анализ пользователей»

---

### 🏷️ Функционал класса `Tags`

В рамках разработанного класса реализован комплекс аналитических инструментов для трансформации хаотичных пользовательских меток в структурированные инсайты. Наш алгоритмический стек решает следующие задачи:

1.  **🧩 Многословные концепты (`most_words`):** Выделяет топ-$n$ тегов с наибольшим количеством отдельных слов. Помогает находить самые детализированные и развернутые описания фильмов.

2.  **📏 Максимальная длина тегов (`longest`):** Измеряет физическую длину меток в символах, возвращая топ-$n$ самых длинных строк, отсортированных по убыванию.

3.  **🔗 Пересечение текстовых гигантов (`most_words_and_longest`):** Находит общие элементы на стыке многословных и длинных тегов. Изолирует уникальные, максимально перегруженные смыслом формулировки.

4.  **🔥 Глас народа (`most_popular`):** Подсчитывает частоту упоминания каждого тега, наглядно демонстрируя главные тренды и доминирующие темы в обсуждениях.

5.  **🔍 Тематический поиск (`tags_with`):** Интеллектуальный фильтр для контекстного поиска уникальных меток по ключевому слову с алфавитной сортировкой.

2.  **🧠 Глубинный срез («Смысловые гиганты» через `most_words` + `longest` + `most_words_and_longest`):**  
    Позволяет заглянуть внутрь пользовательского восприятия. Эти методы изолируют не просто ключевые слова, а уникальные мини-рецензии и сложные концепты. ### В рамках этого класса успешно реализованы следующие аналитические задачи:

🌐 Парсинг данных (get_tmdb): Написан интеллектуальный алгоритм, который заходит на веб-страницы themoviedb, извлекает актуальную информацию (Режиссеров, Бюджеты, Кассовые сборы, Длительность) для каждого фильма и аккуратно сортирует базу по убыванию ID фильмов.

🎬 Топ-режиссёров мирового кино (top_directors): Создан инструмент, выявляющий самых продуктивных и влиятельных создателей кино по количеству снятых шедевров.

💰 Рейтинг киногигантов (most_expensive): Реализован поиск самых финансово затратных и амбициозных кинопроектов в истории.

📈 Анализ абсолютной окупаемости (most_profitable): Рассчитана чистая прибыль картин как разница между мировыми кассовыми сборами и бюджетом, что позволило выявить главных коммерческих триумфаторов индустрии.

⏱️ Измерение кинематографического времени (longest): Выделены самые эпичные и продолжительные киноленты, способные удерживать зрителя у экрана дольше всего.

💎 Индекс плотности инвестиций (top_cost_per_minute): Разработана уникальная метрика «цена минуты», которая показывает, сколько стоила каждая секунда экранного времени для продюсеров, и определяет самые дорогие в производстве фильмы с точностью до двух знаков после запятой.
